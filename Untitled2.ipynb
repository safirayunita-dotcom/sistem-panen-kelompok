```python
%%writefile input_panen.py

def input_data_panen():
    komoditas = input("Nama komoditas: ")
    jumlah_karung = int(input("Jumlah karung: "))
    berat_per_karung = float(input("Berat per karung (kg): "))

    return komoditas, jumlah_karung, berat_per_karung


komoditas, jumlah_karung, berat_per_karung = input_data_panen()

print("\n=== DATA PANEN ===")
print("Komoditas:", komoditas)
print("Jumlah karung:", jumlah_karung)
print("Berat per karung:", berat_per_karung, "kg")
```
